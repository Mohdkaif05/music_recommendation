```python

import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
from sklearn.neighbors import NearestNeighbors



df = pd.read_csv("audio_features_updated.csv")

X = df.iloc[:,:80]
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

pca = PCA(n_components=0.95, random_state=42).fit(X_scaled)
X_pca = pca.transform(X_scaled)

kmeans = KMeans(n_clusters=6, random_state=42)
df['cluster'] = kmeans.fit_predict(X_pca)

cluster_knn_models = {}

for c in range(6):
    cluster_data = X_pca[df['cluster'] == c]
    knn = NearestNeighbors(n_neighbors=6, metric='cosine')
    knn.fit(cluster_data)
    cluster_knn_models[c] = knn

print(f"✅ Model trained with {6} clusters.")

```

```python
import pickle

model_bundle = {
    "scaler": scaler,
    "pca": pca
}

# Save it as a pickle file
with open("scaler_pca.pkl", "wb") as f:
    pickle.dump(model_bundle, f)

print("✅ Model saved successfully as 'audio_recommendation_model.pkl'")
```

```python

with open("kmeans.pkl","wb") as f:
    pickle.dump(kmeans,f)
```

```python

with open("knn.pkl","wb") as f:
    pickle.dump(cluster_knn_models,f)
```
